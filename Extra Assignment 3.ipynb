```python
# Catagory the people by their age like children, adult, citizen, senior citizen...
age = int(input("Enter your age:"))
if(age <= 16):
    print("Children")
elif(age >= 17 and age <= 30):
    print("Adult")
elif(age >= 31 and age <= 59):
    print("Citizen")  
else:
    print("Senior Citizen")
```
